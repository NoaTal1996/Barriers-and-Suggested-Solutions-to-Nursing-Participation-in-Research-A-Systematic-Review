```python
PATH = "/content/drive/MyDrive/fix_systematic_review/Screening_2/output_review_systematic_label.xlsx"
df = pd.read_excel(PATH)
```

```python
df = df[df['Extracted_decision_sytematic'] == 'No']
```

```python
path = '/content/drive/MyDrive/fix_systematic_review/papers_2'
```

```python
download_dir = Path(path)

# Load the CSV file containing the DOIs
pdf_files = list(download_dir.glob('*.pdf'))
filenames = [os.path.splitext(os.path.basename(path))[0] for path in pdf_files]
cleaned_filenames = [filename.replace("_2F", "/") for filename in filenames]
# cleaned_filenames = [filename.replace("_", " ") for filename in cleaned_filenames]
cleaned_filenames = [filename.replace(":", "") for filename in cleaned_filenames]

cleaned_filenames = [filename.replace("_28", "(").replace("_29", ")") for filename in cleaned_filenames]
```

```python
df['Matching_Filename'] = None
cleaned_filenames = [str(i) for i in cleaned_filenames]
# Iterate over the DataFrame rows
for index, row in df.iterrows():

    # Check each of the possible ID columns for matches
    for col in ['DOI', 'PubMed ID', 'Title']:
        if col == 'PubMed ID':
            cleaned_title = str(row[col]).split('.')[0].strip()

        elif col == 'Title':


            cleaned_title = row[col].lower().replace(":", "")
            cleaned_title = cleaned_title.lower().replace("?", "")
            cleaned_title = cleaned_title.lower().replace(".", "")
        else:
            cleaned_title = row[col]
        if isinstance(cleaned_title, str):
            temp = [s.lower() for s in cleaned_filenames]
            if cleaned_title in temp:
                df.at[index, 'Matching_Filename'] = row[col]
        if row[col] in cleaned_filenames:
            df.at[index, 'Matching_Filename'] = row[col]
            break  # Stop checking once a match is found to avoid overwriting with None

# Display the updated DataFrame
# df
```

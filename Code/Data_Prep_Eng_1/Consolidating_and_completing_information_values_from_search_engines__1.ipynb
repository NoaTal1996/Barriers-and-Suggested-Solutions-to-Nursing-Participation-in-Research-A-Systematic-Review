##Completing information from PubMed

```python

from pymed import PubMed
from pymed.article import PubMedArticle
import os
from tqdm import tqdm


def prepare_data_for_pubmed(pubmed_df, output_dir, output_name='pubmed_final.csv'):
    """
    Prepares and cleans the PubMed data, then exports it to a CSV file.
    Fetches abstracts and keywords from the PubMed API for articles with missing data.

    Parameters:
    file_path (str): Path to the PubMed data file (CSV or Excel).
    output_dir (str): Directory where the cleaned CSV will be saved.
    output_name (str, optional): The name of the output cleaned CSV file. Default is "pubmed_final.csv".

    Returns:
    None

    Outputs:
    Saves the cleaned DataFrame as a CSV file in the specified output directory.
    """
    print("Preparing PubMed data...")
    pubmed_features = ["Title", "Publication Year", "DOI", "Journal/Book", "PMID", "Authors"]


    # Initialize PubMed client
    pubmed = PubMed(tool="MyTool", email="noatal96@gmail.com")

    # Filter out rows with null PMIDs
    pubmed_df = pubmed_df[pubmed_df['PMID'].notnull()]

    # Initialize abstracts and keywords with placeholders
    abstracts = [""] * len(pubmed_df)
    keywords = [""] * len(pubmed_df)

    list_of_pubmed_ids = pubmed_df['PMID'].tolist()

    for i in tqdm(range(0, len(list_of_pubmed_ids), 100), desc="Fetching Articles", unit="batch"):

        max_id = min(i + 100, len(list_of_pubmed_ids))
        try:
            articles = pubmed._getArticles(list_of_pubmed_ids[i:max_id])
            for j, article in enumerate(articles):
                index = i + j  # Calculate the actual index in the DataFrame
                abstracts[index] = article.abstract if article.abstract else ""
                # Check if the article is a PubMedArticle and has keywords
                if isinstance(article, PubMedArticle) and hasattr(article, 'keywords'):
                    keywords[index] = str(article.keywords)
                else:
                    keywords[index] = ""  # Set keywords to empty string if not available
        except Exception as e:
            print(f"Error fetching articles for IDs {list_of_pubmed_ids[i:max_id]}: {str(e)}")

    pubmed_df["Abstract"] = abstracts
    pubmed_df["Keywords"] = keywords
    # pubmed_df = pubmed_df[pubmed_df['Abstract'] != '']  # Filter out empty abstracts if necessary

      # Data cleaning and renaming columns
      # pubmed_df['Authors'] = pubmed_df['Authors'].apply(preprocess_authors_list, sep=',')
      # pubmed_df = pubmed_df.rename(columns={'Journal/Book': 'Source_Title', 'Publication Year': 'Year'})
      # pubmed_df_to_export = pubmed_df[features_to_use]

      # Exporting DataFrame
    output_path = os.path.join(output_dir, output_name)
    pubmed_df.to_csv(output_path, index=False, encoding="utf-8")
```
